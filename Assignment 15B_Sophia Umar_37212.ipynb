```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv('student_tea_snacks _1_.csv')

print(df.head())
print(df.info())
```

```python
print(df['tea_type'].value_counts(dropna=False))
```

```python
cups_by_gender = df.groupby('gender')['cups_per_day'].agg(
    ['count', 'mean', 'median', 'min', 'max', 'std']
)
print(cups_by_gender)
```

```python
snack_counts = df['snack_choice'].value_counts()
print(snack_counts)

plt.figure(figsize=(8, 5))
sns.barplot(
    x=snack_counts.index,
    y=snack_counts.values,
    palette='Set2',
    hue=snack_counts.index,
)
plt.title('Student Snack Choices')
plt.xlabel('Snack Choice')
plt.ylabel('Number of Students')
plt.tight_layout()
plt.show()
```

```python
print(pd.crosstab(df['age_group'], df['tea_type'], margins=True))
print(
    (pd.crosstab(df['age_group'], df['tea_type'], normalize='index') * 100).round(
        1
    )
)
```

```python
zero_cups = df[df['cups_per_day'] == 0]
print('Students with 0 cups/day:', len(zero_cups))

no_pref = df[df['prefers_tea'] == 'No']
print("Students with prefers_tea == 'No':", len(no_pref))
```
